# Short-Term Memory (Checkpointers)

Without a checkpointer, each `invoke` is independent. With `InMemorySaver` + a `thread_id`, the agent remembers earlier turns in that thread.

**Key learnings**
- Stateless agents forget prior messages between calls
- `checkpointer=InMemorySaver()` stores conversation state in process memory
- `config={"configurable": {"thread_id": "..."}}` selects which thread to continue


```python
from dotenv import load_dotenv

load_dotenv()
```

## No memory

Two separate invokes: the second question has no access to the name/colour from the first.


```python
from langchain.agents import create_agent


agent = create_agent(
    "gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is SHYAM and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## Memory with InMemorySaver

Attach a checkpointer and reuse the same `thread_id` so follow-up questions retain context.


```python
from langgraph.checkpoint.memory import InMemorySaver

# Persist message history across invokes for a given thread_id
agent = create_agent(
    "gpt-5-nano",
    checkpointer=InMemorySaver(),
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is SHYAM and my favourite colour is Orange")
# thread_id keys the saved conversation
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
